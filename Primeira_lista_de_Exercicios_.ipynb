```python
# Questão 4 - Par ou ímpar

numero = int(input("Digite um número: "))

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

```python
# Questão 5 - Situação do aluno

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print("Média:", media)
```

```python
# Questão 6 - Maior entre três números

a = int(input("Primeiro número: "))
b = int(input("Segundo número: "))
c = int(input("Terceiro número: "))

maior = a

if b > maior:
    maior = b

if c > maior:
    maior = c

print("Maior número:", maior)
```

```python
# Questão 7 - Contagem crescente

n = int(input("Digite N: "))

for i in range(1, n + 1):
    print(i)
```

```python
# Questão 8 - Somando de 1 até N

n = int(input("Digite N: "))

soma = 0

for i in range(1, n + 1):
    soma = soma + i

print("Soma:", soma)
```

```python
# Questão 9 - Contando números pares

pares = 0

for i in range(10):
    numero = int(input("Digite um número: "))

    if numero % 2 == 0:
        pares = pares + 1

print("Foram digitados", pares, "números pares.")
```

```python
# Questão 10 - Soma apenas dos positivos

soma = 0

for i in range(8):
    numero = int(input("Digite um número: "))

    if numero > 0:
        soma = soma + numero

print("Soma dos positivos:", soma)
```

```python
# Questão 11 - Maior valor digitado

maior = None

for i in range(6):
    numero = int(input("Digite um número: "))

    if maior is None or numero > maior:
        maior = numero

print("Maior valor:", maior)
```

```python
# Questão 12 - Tabuada

numero = int(input("Digite um número: "))

for i in range(1, 11):
    resultado = numero * i
    print(numero, "x", i, "=", resultado)
```

```python
# Questão 13 - Tentativas de senha

senha = input("Digite a senha: ")

while senha != "1234":
    senha = input("Senha incorreta. Digite novamente: ")

print("Acesso permitido!")
```

```python
# Questão 14 - Número secreto

secreto = 7
tentativas = 0

numero = int(input("Tente descobrir o número: "))
tentativas = tentativas + 1

while numero != secreto:
    if numero < secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")

    numero = int(input("Tente novamente: "))
    tentativas = tentativas + 1

print("Parabéns! Você acertou.")
print("Tentativas:", tentativas)
```

```python
# Questão 15 - Analisando uma sequência de números

quantidade = 0
positivos = 0
negativos = 0
soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    quantidade = quantidade + 1
    soma = soma + numero

    if numero > 0:
        positivos = positivos + 1
    else:
        negativos = negativos + 1

    numero = int(input("Digite um número: "))

print("Quantidade:", quantidade)
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Soma:", soma)
```
