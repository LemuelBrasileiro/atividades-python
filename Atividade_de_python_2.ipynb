```python
#Questão 04: Estacionamento

horas = int(input("Digite a quantidade de horas: "))

if horas <= 2:
    valor = 8.00
elif horas <= 5:
    valor = 8.00 + (horas - 2) * 3.00
else:
    valor = 17.00 + (horas - 5) * 2.00

print(f"Valor a pagar: R$ {valor:.2f}")
```

```python
#Questão 05: Máquina de venda de ingresso

preco = 25
total = 0

while total < preco:
    valor = int(input("Insira um valor: "))
    total += valor

troco = total - preco

print(f"Total inserido: R$ {total}")
print(f"Troco: R$ {troco}")
```

```python
#Questão 06: Maior valor e sua posição

maior = None
posicao = 0

for i in range(1, 9):
    numero = int(input(f"Digite o {i}º número: "))

    if maior is None or numero > maior:
        maior = numero
        posicao = i

print(f"Maior valor: {maior}")
print(f"Primeira posição: {posicao}")
```

```python
#Questão 07: Segundo maior valor

maior = None
segundo = None

for i in range(1, 9):
    numero = int(input(f"Digite o {i}º número: "))

    if maior is None or numero > maior:
        segundo = maior
        maior = numero
    elif segundo is None or numero > segundo:
        segundo = numero

print(f"Maior: {maior}")
print(f"Segundo maior: {segundo}")
```

```python
#Questao 08: Detectando repetição consecutiva

teve_iguais = False
anterior = None

for i in range(1, 9):
    numero = int(input(f"Digite o {i}º número: "))

    if anterior is not None and numero == anterior:
        teve_iguais = True

    anterior = numero

if teve_iguais:
    print("Sim, foram digitados dois valores iguais consecutivamente.")
else:
    print("Não foram digitados valores iguais consecutivamente.")
```

```python
#Questão 09: Controle de saldo

saldo = 500.00

for i in range(1, 7):
    tipo = int(input(f"Operação {i} - Tipo (1 = depósito, 2 = saque): "))
    valor = float(input(f"Operação {i} - Valor: "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo final: R$ {saldo:.2f}")
```

```python
#Questão 10: Caixa eletronico simplificado

valor = int(input("Digite o valor inteiro positivo: "))

cedulas = [100, 50, 20, 10, 5, 2, 1]
quantidade = 0

for cedula in cedulas:
    qtd = valor // cedula
    quantidade += qtd
    valor %= cedula

print(f"Menor quantidade de cédulas: {quantidade}")
```

```python
#Questão 11: Número primo

n = int(input("Digite um inteiro maior que 1: "))

primo = True

for i in range(2, n):
    if n % i == 0:
        primo = False
        break

if primo:
    print("É primo")
else:
    print("Não é primo")
```

```python
#Questão 12: Divisores de um número

n = int(input("Digite um inteiro positivo: "))

quantidade = 0

print("Divisores:", end=" ")
for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
        quantidade += 1

print()
print(f"Quantidade de divisores: {quantidade}")
```

```python
#Questão 13: Números primos em um intervalo

a = int(input("Digite A: "))
b = int(input("Digite B: "))

quantidade = 0

print("Primos:", end=" ")
for n in range(a, b + 1):
    if n > 1:
        primo = True
        for i in range(2, n):
            if n % i == 0:
                primo = False
                break
        if primo:
            print(n, end=" ")
            quantidade += 1

print()
print(f"Quantidade de primos encontrados: {quantidade}")
```

```python
#Questão 14: Jogo com limites de tentativas

secreto = 37
tentativas = 0
max_tentativas = 5
acertou = False

while tentativas < max_tentativas and not acertou:
    palpite = int(input("Digite seu palpite: "))
    tentativas += 1

    if palpite == secreto:
        print("Parabéns! Você acertou!")
        acertou = True
    elif palpite < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")

if not acertou:
    print("Suas tentativas acabaram. O número secreto era 37.")
```
